```python

# Mini Project: Palindrome Checker
# Using Stack and Queue

from collections import deque

def palindrome_checker():
    text = input("Enter a word or number: ")

    # Convert to lowercase and remove spaces
    text = text.lower().replace(" ", "")

    # Create Stack and Queue
    stack = []
    queue = deque()

    # Add characters to Stack and Queue
    for ch in text:
        stack.append(ch)
        queue.append(ch)

    is_palindrome = True

    # Compare characters
    while stack:
        if stack.pop() != queue.popleft():
            is_palindrome = False
            break
    # Display result
    if is_palindrome:
        print("It is a Palindrome!")
    else:
        print("It is Not a Palindrome!")

# Run the program
palindrome_checker()
```
